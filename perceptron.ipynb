```python
import numpy as np

# les 4 combinaisons possibles des capteurs
X = np.array([[0,0], [0,1], [1,0], [1,1]])
# la réponse voulue pour NOR: 1 si les capteurs sont à 0 sinon 0
y = np.array([1,0,0,0])

# réglages nécessaires pour le calcul
np.random.seed(42)
w = np.random.randn(2)
b = 0.0

# fonction d'activation
def sigmoid(z):
  return 1 / (1 + np.exp(-z))

# c'est le coeur du neurone avec la multiplication des données par le poids
# + le biais
def predict(X,w,b):
  return sigmoid(X @ w + b)

# calcul de l'entropie croisée binaire
def bce(y, p, eps=1e-12):
  return -np.mean(y * np.log(p+ eps) + (1 - y) * np.log(1 - p + eps))

# boucle d'entrainement
lr, epochs = 0.5, 2000
for i in range(epochs):
  # le neurone fait une prédiction avec ses réglages actuels
  p = predict(X,w,b)
  # calcul de l'écart entre la prediction et la vraie réponse (gradient)
  dz = p - y
  # le neurone corrige ses réglages
  w -= lr * (X.T @ dz / len(X))
  b -= lr * dz.mean()
  # le suivi :
  if i % 200 == 0:
    # affiche du code erreur Loss et predictions toutes les 200 itérations
      print(f"Epoch {i}, loss={bce(y,p):.4f}")
print("Prédictions :", predict(X, w, b).round(3))












```
